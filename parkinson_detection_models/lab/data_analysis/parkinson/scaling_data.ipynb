```python
import pandas as pd
import os
from sklearn.preprocessing import StandardScaler
os.chdir("c:/Users/matts/Desktop/wakeup-ai")

df = pd.read_csv("data/parkinson/istanbul_updrs/train_data.csv")
df.head()
```

```python
df.columns
```

```python
acoustic_features = [
    'Jitter(%)', 'Jitter(Abs)', 'Jitter:RAP', 'Jitter:PPQ5', 'Jitter:DDP',
    'Shimmer', 'Shimmer(dB)', 'Shimmer:APQ3', 'Shimmer:APQ5', 'Shimmer:APQ11', 
    'Shimmer:DDA', 'NHR', 'HNR', 'RPDE', 'DFA', 'PPE'
]
```

```python
scaler = StandardScaler()
df[acoustic_features] = scaler.fit_transform(df[acoustic_features])
```

```python
df.to_csv("data/parkinson/istanbul_updrs/processed_istanbul_updrs_data_scaled.csv", index=False)
```
